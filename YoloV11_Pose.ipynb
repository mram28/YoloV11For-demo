```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
import numpy as np

# Define the model path
model_path = r"C:\Users\Malakit Ram\ultralytics\models\V11\Pose\yolo11s-pose.pt"

# Load the YOLO pose model
print("Loading pose model...")
model = YOLO(model_path)
print("Pose model loaded successfully!")

# Open the camera
cap = cv2.VideoCapture(0)  # Change the index if another camera is used

if not cap.isOpened():
    print("Error: Could not open the camera.")
    exit()

# Set the camera resolution (capture resolution from the camera)
capture_width = 1000
capture_height = 720
cap.set(cv2.CAP_PROP_FRAME_WIDTH, capture_width)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, capture_height)

# Set the display frame size (larger frame for visualization)
display_width = 1000  # Larger width for display
display_height = 720  # Larger height for display

print("Starting video stream. Press 'q' to quit.")

use_matplotlib = False  # fallback flag

while True:
    ret, frame = cap.read()
    if not ret:
        print("Error: Could not read frame.")
        break

    # Perform pose detection using the model
    results = model.predict(source=frame, task='pose', show=False)

    # Annotate the frame with pose results
    annotated_frame = frame
    for result in results:
        annotated_frame = result.plot()

    # Resize for display
    larger_frame = cv2.resize(annotated_frame, (display_width, display_height))

    try:
        # Try OpenCV window display
        cv2.imshow("YOLOv11 Pose Estimation", larger_frame)
        key = cv2.waitKey(1) & 0xFF
        if key == ord('q'):
            break
    except cv2.error:
        # Fallback: use matplotlib if imshow fails
        if not use_matplotlib:
            print("\n⚠️ OpenCV GUI not supported. Switching to matplotlib display mode.")
            use_matplotlib = True

        plt.imshow(cv2.cvtColor(larger_frame, cv2.COLOR_BGR2RGB))
        plt.axis('off')
        plt.show(block=False)
        plt.pause(0.01)
        plt.clf()

# Release resources
cap.release()
cv2.destroyAllWindows()
plt.close()
```
