```python
import cv2
from ultralytics import YOLO

# Define the model path (OBB model)
model_path = r"C:\Users\Malakit Ram\ultralytics\models\V11\OBB\yolo11n-obb.pt"

# Load the YOLO OBB model
print("Loading OBB model...")
model = YOLO(model_path)
print("OBB model loaded successfully!")

# Open the camera
cap = cv2.VideoCapture(0)  # Change index if another camera is used

if not cap.isOpened():
    print("Error: Could not open the camera.")
    exit()

# Set desired frame width and height for capturing
capture_width = 1000
capture_height = 720
cap.set(cv2.CAP_PROP_FRAME_WIDTH, capture_width)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, capture_height)

print("Starting video stream. Press 'q' to quit.")
while True:
    ret, frame = cap.read()
    if not ret:
        print("Error: Could not read frame.")
        break

    # Let the model infer it's an OBB model from the weights
    results = model.predict(source=frame, show=False, verbose=False)

    for result in results:
        annotated_frame = result.plot()  # draws oriented boxes for OBB models
        display_frame = cv2.resize(annotated_frame, (capture_width, capture_height))
        cv2.imshow("YOLOv11 OBB Detection", display_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
