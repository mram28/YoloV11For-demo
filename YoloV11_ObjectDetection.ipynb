```python
import cv2
from ultralytics import YOLO

# Define the model path
model_path = r"C:\Users\Malakit Ram\ultralytics\models\V11\ObjectDetection\yolo11s.pt"

# Load the YOLO object detection model
print("Loading object detection model...")
model = YOLO(model_path)
print("Object detection model loaded successfully!")

# Open the camera
cap = cv2.VideoCapture(0)  # Change the index if another camera is used

if not cap.isOpened():
    print("Error: Could not open the camera.")
    exit()

# Set desired frame width and height for capturing
capture_width = 1000
capture_height = 720
cap.set(cv2.CAP_PROP_FRAME_WIDTH, capture_width)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, capture_height)

# Process video frames
print("Starting video stream. Press 'q' to quit.")
while True:
    ret, frame = cap.read()
    if not ret:
        print("Error: Could not read frame.")
        break

    # Perform object detection using the model
    results = model.predict(source=frame, task='detect', show=False)

    # Iterate through the results and annotate the frame
    for result in results:
        annotated_frame = result.plot()  # Plot the object detection results

        # Resize the annotated frame for display
        display_frame = cv2.resize(annotated_frame, (capture_width, capture_height))

        # Display the frame
        cv2.imshow("YOLOv11 Object Detection", display_frame)

    # Exit loop when 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
```
